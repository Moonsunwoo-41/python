```python
# 집합
import random
import requests

def fetch_lotto_info(): 
    URL = 'https://www.dhlottery.co.kr/lt645/selectPstLt645Info.do'
    res = requests.get(URL)
    raw_data = res.text
    data = res.json()
    core_data = data['data']['list'][0]
    lucky = []
    for k, v in core_data.items(): # key, value 세팅법
        if 'tm' in k:
            lucky.append(v)
    bonus = core_data['bnsWnNo']
    return lucky, bonus

def auto_lotto():
    '''자동 로또 번호 생성 및 당첨 확인 함수'''
    
    ran_ball = random.sample(range(1,46), 7)
    ran_tm = sorted(ran_ball[:6])
    ran_bns = ran_ball[6]
    bns_correct = False
    tm_correct = len(set(ran_tm) & set(lucky))

    if ran_bns == bonus:
        bns_correct = True
    
    if tm_correct == 6:
        rank = 1
    elif tm_correct == 5:
        if bns_correct:
            rank = 2
        else:
            rank = 3
    elif tm_correct == 4:
        rank = 4
    elif tm_correct == 3:
        rank = 5
    else:
        rank = '꽝'

    return f'자동 번호: {ran_tm}, 자동 보너스: {ran_bns}, 등수: {rank}'
    
print(fetch_lotto_info())    
# auto_lotto([2, 13, 18, 32, 38, 42], 22)  
auto_lotto()


```

```python
BASE_URL = 'https://naverapihub.apigw.ntruss.com'
NEWS_URL = '/search/v1/news'

import requests
# 인증 관련 헤더
headers = {
    'X-NCP-APIGW-API-KEY-ID': NAVER_CLIENT_ID,
    'X-NCP-APIGW-API-KEY': NAVER_CLIENT_SECRET
}

URL = BASE_URL + NEWS_URL

# 쿼리 파라미터를 dict 로 작성
params = {
    'query': '엔화',
    'sort': 'sim',
    'display': 100
}

res = requests.get(URL, headers=headers, params=params)

news = []
items = res.json()['items']
for n in range(params['display']):
    clean_title = items[n]['title'].replace('<b>', '').replace('</b>', '')
    link = items[n]['link']
    if 'naver' in link:
        news.append({'title': clean_title, 'link': link})
         
news

import csv

# 컬럼명(헤더)으로 사용할 키값 지정
fieldnames = ['title', 'link']

# 파일 열기 (한글 깨짐 방지 utf-8-sig, 줄바꿈 방지 newline='')
with open('news_result.csv', mode='w', encoding='utf-8-sig', newline='') as f:
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    
    # 1. 헤더(컬럼명) 쓰기
    writer.writeheader()
    
    # 2. 데이터(딕셔너리 리스트) 쓰기
    writer.writerows(news)
```
