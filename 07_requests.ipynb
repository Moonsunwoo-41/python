```python
def extract_naver_news(URL):
    if 'n.news.naver.com' not in URL:
        raise Exception('네이버뉴스가 아닙니다.')
        
    res = requests.get(URL)
    soup = BeautifulSoup(res.text, 'html.parser')
    return soup.select_one('#dic_area').text.strip()

extract_naver_news('https://n.news.naver.com/article/008/0005405342')

```

```python

BASE_URL = 'https://naverapihub.apigw.ntruss.com'
NEWS_URL = '/search/v1/news'

import requests
from bs4 import BeautifulSoup
# 인증 관련 헤더
headers = {
    'X-NCP-APIGW-API-KEY-ID': NAVER_CLIENT_ID,
    'X-NCP-APIGW-API-KEY': NAVER_CLIENT_SECRET
}

URL = BASE_URL + NEWS_URL

news = []

def content_naver_news(q: str):
    params = {
    'query': q,
    'sort': 'sim',
    'display': 5
    }
    res = requests.get(URL, headers=headers, params=params)
    items = res.json()['items']
    
    for n in range(params['display']):
        link = items[n]['link']
        clean_title = items[n]['title'].replace('<b>', '').replace('</b>', '')
        
        soup_res = requests.get(link)
        soup = BeautifulSoup(soup_res.text, 'html.parser')
        content = soup.select_one('#dic_area').text.strip()
        
        news.append({
            'title': clean_title, 
            'link': link, 
            'content': content})
    
    return news


```
