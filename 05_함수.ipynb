````python
# name: plus_one / in: 숫자 / out: 숫자 + 1
def plus_one(n):
    return n + 1

plus_one(5)
````

````python
# n번 인사하는 함수
# name: hello_ntimes

def hello_ntimes(n):
    for _ in range(n):
        print('안녕')
        
hello_ntimes(10)
    
````

````python
# 3제곱 해주는 함수
# name: cube, in: 숫자 1개, out: 숫자의 세제곱
def cube(n):
    return n ** 3

cube(5)
````

````python
# 원기둥의 부피를 구하는 함수
# name: cylinder, in: 반지름/높이, out: 부피(pi * r^2 * h)
import math

def cylinder(r, h):
    '''원기둥의 부피를 구합니다.
    r -> 반지름
    h -> 높이
    '''
    
    return math.pi * r ** 2 * h

print(cylinder(3, 2))
````

````python
# 리스트 내부의 숫자들의 평균을 구해주는 함수
# name: get_avg, in: 숫자들로 만들어진 리스트, out: 평균
def get_avg(numbers):
    '''리스트 내부의 숫자들의 평균을 구해주는 함수
    
    numbers: 숫자들로 만들어진 리스트, 
    output: 평균'''
    
    return sum(numbers) / len(numbers)

get_avg([3, 2, 1, 4, 5])
````

````python
# 비밀번호 체크해주는 함수 (숫자 포함, 8글자 이상)
# name: check_pw, in: 비밀번호 문자열, out: True/False

def check_pw(pw):
    '''비밀번호 체크해주는 함수
    pw: 비밀번호(문자열)'''
    
    is_valid_pw = False
    
    if len(pw) >= 8:
        for char in pw:
            if char in '0123456789':
                is_valid_pw = True
                break
    
    return is_valid_pw

check_pw('afr4')
````

````python
# BMI 기준으로, 비만여부 확인해주는 함수 (BMI = 체중(kg) / 키(m)^2)
# 남성 27 이상일 경우 비만
# 여성 25 이상일 경우 비만

def check_obs(height, weight, gender):
    '''BMI 기준으로, 비만여부 확인해주는 함수 
    (BMI = 체중(kg) / 키(m)^2)
    
    남성 27 이상일 경우 비만, 
    여성 25 이상일 경우 비만
    gender = m or f'''
    
    if gender == 'm':
        if weight / height ** 2 >= 27:
            return '비만'
        else:
            return '정상'
    else:
        if weight / height ** 2 >= 25:
            return '비만'
        else:
            return '정상'
        
check_obs(2, 100, 'f')
````
