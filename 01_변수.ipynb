## 변수 (Variable)
- 변수는 박스
- 변수 a 에 10 이라는 값을 저장(할당)한다.
- `=` 할당자는 프로그래밍에서는 "우측의 값을 좌측의 변수에 저장한다."


```python
a = 10
a = 20
print(a)
```

```python
# i -> 5, j -> 3 , s -> '파이썬'
i = 5
j = 3
s = '파이썬' 

print(i, j, s)
```

```python
print(i + j)
print(i - j)
```

```python
a = 1
b = 2
# 우측(a+b) 를 좌측(a)에 저장한다
a = a + b

print(a)
```

```python
# 글자끼리 더하기 -> 이어붙이기
s + ' 쉽다'
```

```python
# 동시 할당
x, y, z = 1, 2, 3

print(x, y, z)
```

```python
# 문제 
# x = 10, y = 20 일때, 값을 서로 교체하는 코드를 작성해 보자.

# swap
x, y = 10, 20
x, y = y, x 

print(x, y) 

# tmp
tmp = x
x = y
y = tmp

print(x, y) 

# 이게 더 좋은듯!!
x = [x, y]
y = x[0]
x = x[1]

print(x, y)

# 정답
x = x + y
y = x - y
x = x - y

print(x, y)
 
```

### f-string (`f'hi'`, `f'''hi'''`)
String 안에 변수를 녹여서 써여할 때 사용

```python
name = '마이클'
age = 25

# 마이클님 환영합니다. 마이클님의 등급은 일반이고, 나이는 25으로 확인됩니다.

print(name + '님 환영합니다. ' + name + '님의 등급은 일반이고, 나이는 '+ str(age) + '으로 확인됩니다.')

print(f'{name}님 환영합니다. {name}님의 등급은 일반이고, 나이는 {age}으로 확인됩니다.')
```

```python
# String 안에서 따옴표 사용하기
print('철수가 말하길 \'안녕 반가워\'')
print("철수가 말하길 '안녕 반가워'")
```
