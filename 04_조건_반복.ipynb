```python
# 불합: 0-399
# 1: 400-599
# 2: 600-749
# 3: 750-899
# 4: 900-1000
score = 30
if score >= 900:
    lv = 4
elif score >= 750:
    lv = 3
elif score >= 600:
    lv = 2
elif score >= 400:
    lv = 1
else:
    lv = 'Failed'

print(lv)
```

```python
# 우수 학생 판별
status = {
    'score': 60,
    'absent': 10
}

# 1. 점수가 70을 넘는 학생이면 우수학생
# 2. 우수학생 중에 결석일이 5일 미만이면 최우수학생
# 3. (우수 아닌) 결석일이 5일 미만이면 성실학생
# 4. 나머지는 일반 학생

if status['score'] > 70:
    if status['absent'] < 5:
        print('최우수학생')
    else:
        print('우수학생')
elif status['absent'] < 5:
    print('성실학생')
else:
    print('일반학생')

```

놀이공원 입장료 계산
놀이공원 입장료를 계산하는 프로그램을 작성하시오.

age = 23
is_weekend = True
has_coupon = False
조건은 다음과 같다.

기본 입장료는 30,000원
13세 미만은 15,000원
65세 이상은 20,000원
주말에는 기본 입장료에 5,000원 추가
쿠폰이 있으면 최종 금액에서 5,000원 할인
단, 13세 미만과 65세 이상은 주말 추가 요금이 없다.
최종 입장료를 출력하시오.

```python
age = 23
is_weekend = True
has_coupon = False
if age < 13:
    cost = 15000
elif age >= 65:
    cost = 20000
else:
    cost = 30000
    if is_weekend:
        cost += 5000
if has_coupon:
    cost -= 5000

print(cost)
```

회원 등급 판정
다음과 같은 고객 정보가 있다.

customer = {
    "name": "김민수",
    "age": 25,
    "purchase": 350000,
    "is_member": True
}
고객의 등급을 다음 조건에 따라 판정하시오.

회원이 아니면 "일반"
회원이면서 구매금액이 500,000원 이상 → "VIP"
회원이면서 구매금액이 200,000원 이상 → "GOLD"
그 외 회원 → "SILVER"
결과 예시:

김민수님의 등급은 GOLD입니다.

```python
customer = {
    "name": "김민수",
    "age": 25,
    "purchase": 350000,
    "is_member": True
}

if customer['is_member']:
    if customer['purchase'] >= 500000:
        tier = 'VIP'
    elif customer['purchase'] >= 200000:
        tier = 'GOLD'
    else:
        tier = 'SILVER'
else:
    tier = 'BRONZE'

print(tier)

# 정답 # code 한 줄이 더 줄어듬

if not customer['is_member']:
    tier = 'BRONZE'
elif customer['purchase'] >= 500000:
    tier = 'VIP'
elif customer['purchase'] >= 200000:
    tier = 'GOLD'
else:
    tier = 'SILVER'
    
print(tier)
```

로그인 가능 여부 확인
다음 정보를 이용하여 로그인 가능 여부를 판단하시오.

user_id = "admin"
password = "python123"

blocked_users = ["test", "guest", "admin2"]
조건:

아이디가 "admin"이고 비밀번호가 "python123"이면 로그인 성공
단, blocked_users에 아이디가 포함되어 있으면 로그인 실패
아이디 또는 비밀번호가 틀리면 "아이디 또는 비밀번호가 틀렸습니다."
차단된 사용자는 "차단된 사용자입니다."
결과를 출력하시오.

```python
user_id = "admin"
password = "python123"

blocked_users = ["test", "guest", "admin2"]

if user_id in blocked_users:
    msg = '차단된 사용자입니다.'
elif user_id == 'admin' and password == 'python123':
    msg = '로그인에 성공하셨습니다.'
else:
    msg = '아이디 혹은 비밀번호가 틀렸습니다.'
    
print(msg)
```

배송비 계산
온라인 쇼핑몰의 주문 정보가 다음과 같다.

order = {
    "price": 48000,
    "is_member": True,
    "region": "서울",
    "coupon": "WELCOME"
}
배송비를 계산하시오.

조건:

상품 금액이 50,000원 이상이면 무료배송
회원이면 배송비 2,000원
비회원이면 배송비 3,000원
"제주" 지역이면 위 배송비에 3,000원 추가
"WELCOME" 쿠폰을 가지고 있으면 배송비 무료
단, 상품 금액이 10,000원 미만이면 쿠폰이 있어도 배송비 무료가 아니다.

```python
order = {
    "price": 48000,
    "is_member": False,
    "region": "제주",
    "coupon": 'none'
}

if order['is_member']:
    fee = 2000
else:
    fee = 3000

if order['price'] >= 50000:
    fee = 0
    
if order["coupon"] == 'WELCOME' and order["price"] >= 10000:
    fee = 0
    
if order["region"] == '제주':
    fee += 3000

print(fee)


    
```

### `while` 구문

```python
n = 0

while n < 5:
    print('출력', n)
    n += 1
    
print('종료 이후 n:', n)
```

```python
numbers = [1, 2, 3, 4, 5]

total = 0
idx = 0

while idx < len(numbers):
    total += numbers[idx]
    print('중간점검', total)
    idx += 1

print(total)
```

### `for .. in` 구문

```python
numbers = [1, 2, 3, 4, 5]
total = 0

for num in numbers:
    print(f'기존{total}에 {num}를 더합니다.')
    total += num
    
print(total)
```
