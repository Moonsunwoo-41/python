## 점수 평균과 합격 판별

학생 점수 리스트가 있다.

```py
scores = [80, 75, 90, 60, 100]
```
평균을 구하시오.
평균이 80 이상이면 "합격" 아니면 "불합격" 출력

````python
scores = [80, 75, 90, 60, 100]
avg = sum(scores) / len(scores)
print(avg)


if avg >= 80:
    print('합격')
else:
    print('불합격')
````

## 이름 존재 여부 확인
```
names = ["철수", "영희", "민수", "지영"]
```
변수 target 에 저장된 이름이 리스트 안에 있으면
"존재합니다" 아니면 "없습니다" 출력

TIP: `in` 연산자 없이 풀기

````python
target = '영수'
names = ["철수", "영희", "민수", "지영"]
idx = 0
msg = '없습니다.'

# while문
while idx < len(names):
    if names[idx] == target:
        msg = '존재합니다.'
        break
    idx += 1
    
print(msg)

# for문
for name in names:
    if target == name:
        msg = '존재합니다.'
        break
print(msg)
````

## 등급 매기기

점수 리스트:
```py
scores = [95, 82, 67, 74, 88]
```
각 점수에 대해

- 90 이상 → 'A'
- 80 이상 → 'B'
- 70 이상 → 'C'
- 그 외 → 'D'

등급 리스트를 새로 만들어 출력하시오.

````python
scores = [95, 82, 67, 74, 88]
grades = []
idx = 0

for score in scores:
    if score >= 90:
        grades.append('A')
    elif score >= 80:
        grades.append('B')
    elif score >= 70:
        grades.append('C')
    else:
        grades.append('D')
            
print(grades)
````

numbers = [12, 7, 25, 4, 18, 31, 10, 9, 42, 15]
반복문과 조건문을 사용하여 다음을 구하시오.

짝수의 개수
홀수의 개수
10 이상인 숫자의 합
가장 큰 숫자
최종 결과가 다음과 같이 나오도록 하시오.

짝수 개수: 6
홀수 개수: 4

가장 큰 숫자: 42
조건: max(), sum() 사용 금지

````python
numbers = [12, 7, 25, 4, 18, 31, 10, 9, 42, 15]
odd = 0
even = 0
over_ten = 0
max_num = numbers[0]

for n in numbers:
    if n % 2:
        odd += 1
    else:
        even += 1
    if n >= 10:
        over_ten += n
    if n >= max_num:
        max_num = n
        
print(f'''짝수 개수: {even} 
      홀수 개수: {odd}
      10 이상인 숫자의 합: {over_ten}
      가장 큰 숫자: {max_num}''')
        
        
````

숫자 등급 분류하기
다음 숫자 리스트가 주어져 있다.

numbers = [95, 72, 88, 61, 45, 100, 77, 53, 82, 39]
반복문과 조건문을 사용하여 각 숫자를 다음 기준으로 분류하시오.

90 이상 -- "A"
80 이상 -- "B" -70 이상 -- "C"
60 이상 -- "D"
60 미만 -- "F"
그리고 각 등급에 해당하는 숫자의 개수도 구하시오.

예상 출력:

95 -- A
72 -- C
88 -- B
61 -- D
45 -- F
100 -- A
77 -- C
53 -- F
82 -- B
39 -- F

A: 2개
B: 2개
C: 2개
D: 1개
F: 3개

````python
numbers = [95, 72, 88, 61, 45, 100, 77, 53, 82, 39]
a_tier = 0
b_tier = 0
c_tier = 0
d_tier = 0
f_tier = 0

for n in numbers:
    if n >= 90:
        a_tier += 1
        print(f'{n} -- A')
    elif n >= 80:
            b_tier += 1
            print(f'{n} -- B')
    elif n >= 70:
            c_tier += 1
            print(f'{n} -- C')
    elif n >= 60:
            d_tier += 1
            print(f'{n} -- D')
    else:
            f_tier += 1
            print(f'{n} -- F')

print(f'''
      A: {a_tier}개
      B: {b_tier}개
      C: {c_tier}개
      D: {d_tier}개
      F: {f_tier}개
      ''')
````

편의점 매출 분석
편의점에서 판매된 상품 데이터가 다음과 같이 주어져 있다.

sales = [
    {"name": "삼각김밥", "price": 1500, "quantity": 12},
    {"name": "라면", "price": 1200, "quantity": 8},
    {"name": "콜라", "price": 1800, "quantity": 5},
    {"name": "샌드위치", "price": 3500, "quantity": 3},
    {"name": "커피", "price": 2500, "quantity": 10}
]
각 상품의 매출액 = 가격 × 판매수량이라고 한다.

반복문과 조건문을 사용하여 다음을 출력하시오.

각 상품의 매출액
전체 매출액
매출액이 20,000원 이상인 상품의 이름
가장 많이 팔린 상품의 이름
예상 출력:

삼각김밥: 18000원
라면: 9600원
콜라: 9000원
샌드위치: 10500원
커피: 25000원

전체 매출액: 72100원

매출액 20000원 이상:
커피

가장 많이 팔린 상품: 삼각김밥

````python
sales = [
    {"name": "삼각김밥", "price": 1500, "quantity": 12},
    {"name": "라면", "price": 1200, "quantity": 30},
    {"name": "콜라", "price": 1800, "quantity": 5},
    {"name": "샌드위치", "price": 3500, "quantity": 3},
    {"name": "커피", "price": 2500, "quantity": 10}
]
total_sales = 0
main = []
top_sales = 0
top_product = ''

for product in sales:
    print(f'{product['name']}: {product['price'] * product['quantity']}원') # 각 상품의 매출액
    total_sales += product['price'] * product['quantity'] # 전체 매출액
    
    # 매출액이 20,000원 이상인 상품의 이름
    if product['price'] * product['quantity'] >= 2e4:
        main.append(product['name'])
    
    if product['price'] * product['quantity'] > top_sales:
        top_sales = product['price'] * product['quantity']
        top_product = product['name']

print(f'''
전체 매출액: {total_sales}원

매출액 20,000원 이상: {main}

가장 많이 팔린 상품: {top_product}
''')


    
````
