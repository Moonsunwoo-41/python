## 리스트 (List)

```python
numbers = [10, 20, 30, 40, 50]
numbers[0] # 10
numbers[2] # 30
numbers[-1] # 50
numbers[-2] # 40

# indexerror
# numbers[5]
# numbers[-6]

numbers[0] + numbers[1] # 30

#Slicing
numbers[0:3] # [10, 20, 30]
numbers[2:2] # [] empty list

numbers[0:5]
numbers[:]

numbers == numbers[:] # True
numbers is numbers[:] # False
```

```python
# 리스트 수정
numbers = [10, 20, 30, 40, 50]

# 특정 index 의 데이터 바꾸기
numbers[2] = 300
numbers # [10, 20, 300, 40, 50]

# 리스트에 데이터 추가하기
numbers.append(60)
numbers

# 리스트의 데이터 삭제하기
numbers.pop(1)
numbers

l1 = [1, 2, 3]
l2 = [4, 5, 6]

l1 + l2
```

## 범위(Range)

```python
# range 는 처음부터 마지막-1 까지의 정수만을 의미
r1 = range(1,10)
print(list(r1))

# 시작숫자 x -> 0부터 시작
r2 = range(5)
print(list(r2))

# step 개념
r3 = range(0, 10, 3)
print(list(r3))

# 음수 step(뒤로가기)
r4 = range(0, 10 ,-1)
print(list(r4))

r5 = range(10, 0, -1)
print(list(r5))
```

```python
import random

random.choice(['김', '이', '박', '최'])
random.sample(['김', '이', '박', '최'], 3)

x = ['김', '이', '박', '최']
random.shuffle(x)
print(x)

# 로또 추첨기

balls = range(1, 46)
lotto = random.sample(balls, 6)
lotto.sort()
lotto.sort(reverse = True)
print(lotto)
```

## Dict (딕셔너리)
| key - Value
- JSON 의 뜻은 JavaScript Object Notation 의 줄임말
    - JS의 Object자료형 식 표기법 (" 써야함)
- Python 에는 Dict 자료형

```python
# 생성
me = {
    'name': 'moon',
    'age': 23,
    'is_hungry': False,
    'height': 180,
    'skills': ['n8n', 'sql', 'python']
}
me
```

```python
# 접근 -> 대괄호 안에 key 를 쓴다.

#이름, 나이 출력
me['name']
me['age']

#skill 목록 출력
me['skills']

#skill 중 'python' 출력
me['skills'][2]
me['skills'][-1]
```

```python
# Value 변경
me['is_hungry'] = True
me['is_hungry']

me['skills'].append('git')
print(me)
```

```python
# K-V 추가
me['money'] = 10000
me

# K-V 제거
me.pop('height')
me
```

## Set (집합)
- 중복되는 요소가 없다.
- 순서(index)가 없다.

```python
s1 = {1, 1, 2, 3, 3, 3}
s2 = {3, 4, 5, 5}

print(s1, s2)

# 합집합
print(s1 | s2)

# 교집합
print(s1 & s2)

# 차집합
print(s1 - s2)

# 중복 제거방법
x = [1, 2, 3, 3, 3, 3, 2, 1, 2, 1, 2, 2, 3, 3, 2, 1, 1, 3, 3, 1]
uniq_x = list(set(x))
```
